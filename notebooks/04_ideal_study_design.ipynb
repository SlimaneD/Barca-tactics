## The limitation of what we have

The Messi-era analysis (notebook 02) uses full StatsBomb event data with spatial coordinates. The Flick-era analysis (notebook 03) uses match-level aggregates. These two datasets do not speak the same language:

- Messi era: defensive line height from actual event locations
- Flick era: no direct line-height measurement

We cannot put them in the same regression and compare coefficients directly.

---

## What the ideal data would look like

### Option A — StatsBomb 360 (event-linked freeze frames)

StatsBomb 360 attaches visible player positions to key events (passes, shots, pressures). For the offside trap, the critical event is **the pass played over or through the line**. With 360 data we can measure:

- **Defensive line position at the triggering pass**: average x-coordinate of the four defenders at the moment the pass is released
- **Gap behind the line**: distance from last defender to goalkeeper
- **Attacker depth**: how far behind the line the attacker receives (or would receive if not flagged)

This gives a continuous, event-level measure of *how exposed* the line was on each occasion — regardless of whether the trap worked.

### Option B — Continuous tracking data (SkillCorner / Second Spectrum)

Tracking data provides player positions at 25 fps throughout the match. This enables:

- **True defensive line height at every moment**: not just at event moments, but as a time series
- **Line synchronisation**: the standard deviation of the four defenders' x-positions — a fragmented line is more vulnerable
- **Press-to-line ratio over time**: does the team maintain the press while keeping the line high, or do they decouple?
- **Recovery speed**: when a pass breaks the line, how quickly do defenders track back?

---

## The ideal outcome variable

Neither "offsides drawn" nor "goals conceded" is ideal:
- Offsides drawn = trap success, but ignores near-misses where the trap was beaten but the shot was missed
- Goals conceded = confounded by shot quality and goalkeeping

The ideal variable is **xG conceded on behind-the-line chances** — shots that originated from an attacker who was level with or beyond the last defender at the triggering pass. This requires 360 or tracking data to identify the set of such chances, and then applies standard xG to measure their danger.

---

## The cross-era regression

With both eras available in the same format, the model would be:

```
xG_behind_line_per_match ~
    def_line_height          # how high the line sits
  + ppda                     # press intensity (lower = more pressing)
  + line_synchronisation     # SD of defenders' x-positions
  + opponent_directness      # does the opponent play long/direct?
  + era                      # Messi era vs Flick era dummy
  + (1 | season)             # season-level random effect
  + (1 | opponent)           # opponent-level random effect
```

The `era` coefficient answers: *is Flick's Barça more exposed behind the line than previous eras, after controlling for line height and press intensity?*

If `era` is significant and positive, the deterioration is over and above what can be explained by line height and pressing — suggesting execution or personnel effects.

If `era` becomes insignificant when `ppda` and `line_synchronisation` are included, the deterioration is explained by a less coordinated press-and-line combination.
